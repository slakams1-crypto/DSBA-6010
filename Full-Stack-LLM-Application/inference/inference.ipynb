```python
from huggingface_hub import InferenceClient

# Initialize HuggingFace client
hf_client = InferenceClient(token=os.getenv("HUGGINGFACE_API_KEY"))

def huggingface_generate(prompt, model="mistralai/Mistral-7B-Instruct-v0.2", max_tokens=500):
    """
    Generate text using HuggingFace Inference API

    Args:
        prompt: Input text prompt
        model: HuggingFace model ID
        max_tokens: Maximum tokens to generate

    Popular models:
        - mistralai/Mistral-7B-Instruct-v0.2
        - meta-llama/Llama-2-7b-chat-hf
        - tiiuae/falcon-7b-instruct
        - google/flan-t5-xxl
    """
    try:
        response = hf_client.text_generation(
            prompt,
            model=model,
            max_new_tokens=max_tokens,
            temperature=0.7,
            top_p=0.95,
        )
        return response
    except Exception as e:
        return f"Error: {e}"

# Example usage (commented out as mistralai/Mistral-7B-Instruct-v0.2 is a chat model)
prompt = "Explain what makes open-source LLMs important for AI democratization."
response = huggingface_generate(prompt, model="mistralai/Mistral-7B-Instruct-v0.2")
print(f"HuggingFace Response:\n{response}")
```

```python
# Chat completion with HuggingFace
def huggingface_chat(messages, model="mistralai/Mistral-7B-Instruct-v0.2", max_tokens=500):
    """
    Chat using HuggingFace Inference API
    """
    try:
        response = hf_client.chat_completion(
            messages=messages,
            model=model,
            max_tokens=max_tokens,
        )
        return response.choices[0].message.content
    except Exception as e:
        return f"Error: {e}"

# Example chat
messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "What are the benefits of using open-source LLMs?"}
]

response = huggingface_chat(messages)
print(f"HuggingFace Chat Response:\n{response}")
```

```python
# Streaming with HuggingFace

def huggingface_stream(prompt, model="mistralai/Mistral-7B-Instruct-v0.2"):
    """
    Stream responses from HuggingFace
    """
    try:
        print("Streaming response:")
        full_response = ""

        for token in hf_client.text_generation(
            prompt,
            model=model,
            max_new_tokens=500,
            stream=True
        ):
            print(token, end='', flush=True)
            full_response += token

        print()  # New line
        return full_response
    except Exception as e:
        return f"Error: {e}"

# Example streaming
huggingface_stream("List the key components of a transformer model.")
```

```python
import pkgutil

def list_submodules(package_name):
    try:
        package = __import__(package_name)
        print(f"\nSubmodules of {package_name} (containing 'memory'):")
        found_memory_module = False
        for _, modname, _ in pkgutil.walk_packages(package.__path__, package.__name__ + '.'):
            if "memory" in modname:
                print(modname)
                found_memory_module = True
        if not found_memory_module:
            print(f"No 'memory' related submodules found in {package_name}")
    except ImportError:
        print(f"Package '{package_name}' not found.")

list_submodules('langchain')
list_submodules('langchain_community')
```

```python
import time

def compare_models(prompt, models_dict):
    """
    Compare responses from different models

    Args:
        prompt: The input prompt
        models_dict: Dictionary of {name: llm_instance}
    """
    print(f"Prompt: {prompt}\n")
    print("="*80)

    results = {}

    for name, llm in models_dict.items():
        print(f"\n{name}:")
        print("-" * 80)

        try:
            start_time = time.time()
            response = llm.invoke(prompt)
            end_time = time.time()

            response_text = response.content if hasattr(response, 'content') else str(response)

            print(response_text)
            print(f"\nTime taken: {end_time - start_time:.2f} seconds")

            results[name] = {
                "response": response_text,
                "time": end_time - start_time
            }
        except Exception as e:
            print(f"Error: {e}")
            results[name] = {"error": str(e)}

    return results

# Example comparison
models_to_compare = {
    #"GPT-3.5": openai_llm,
    #"Claude": claude_llm,
    "Ollama": ollama_llm  # Uncomment if Ollama is running
}

prompt = "Explain the concept of 'hallucination' in LLMs in 2 sentences."
comparison_results = compare_models(prompt, models_to_compare)
```

```python
# Approximate pricing (as of 2024 - check current pricing)
PRICING = {
    "gpt-4": {"input": 0.03, "output": 0.06},  # per 1K tokens
    "gpt-3.5-turbo": {"input": 0.0005, "output": 0.0015},
    #"claude-3-opus": {"input": 0.015, "output": 0.075},
    #"claude-3-5-sonnet": {"input": 0.003, "output": 0.015},
    #"claude-3-haiku": {"input": 0.00025, "output": 0.00125},
    #"gemini-1.5-pro": {"input": 0.00125, "output": 0.005},
    #"gemini-1.5-flash": {"input": 0.000075, "output": 0.0003},
}

def estimate_cost(model, input_tokens, output_tokens):
    """
    Estimate API cost

    Args:
        model: Model name
        input_tokens: Number of input tokens
        output_tokens: Number of output tokens
    """
    if model not in PRICING:
        return "Pricing not available for this model"

    input_cost = (input_tokens / 1000) * PRICING[model]["input"]
    output_cost = (output_tokens / 1000) * PRICING[model]["output"]
    total_cost = input_cost + output_cost

    return {
        "input_cost": f"${input_cost:.6f}",
        "output_cost": f"${output_cost:.6f}",
        "total_cost": f"${total_cost:.6f}"
    }

# Example cost estimation
print("Cost for 1000 input tokens and 500 output tokens:")
print("\nGPT-3.5-Turbo:", estimate_cost("gpt-3.5-turbo", 1000, 500))
print("GPT-4:", estimate_cost("gpt-4", 1000, 500))
#print("Claude 3.5 Sonnet:", estimate_cost("claude-3-5-sonnet", 1000, 500))
#print("Gemini 1.5 Flash:", estimate_cost("gemini-1.5-flash", 1000, 500))
```
