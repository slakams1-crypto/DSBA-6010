````python
from ollama import chat
import json

def test_prompt(prompt, test_cases):
    results = []
    for case in test_cases:
        response = chat(
        model="llama3",
        messages=[{"role": "user", "content": prompt.format(**case)}],
        temperature=0
    )
    results.append({
            "input": case,
            "output": response.choices[0].message.content,
            "valid_json": is_valid_json(response.choices[0].message.content)
    })
    return results
````

````python
def evaluate_llm_output(original_prompt, llm_response, objective, evaluation_model="llama3"):
    """
    Evaluates if an LLM's response meets a specified objective using another LLM.

    Args:
        original_prompt (str): The prompt given to the original LLM.
        llm_response (str): The response generated by the original LLM.
        objective (str): The objective that the llm_response should meet.
        evaluation_model (str): The model to use for the evaluation.

    Returns:
        str: A string indicating whether the objective was met and why.
    """
    evaluation_prompt = f"""
    You are an expert evaluator. Your task is to determine if an AI's response meets a specific objective.

    Original Prompt: '''{original_prompt}'''
    AI Response: '''{llm_response}'''
    Objective: '''{objective}'''

    Based on the above, does the AI Response completely and accurately fulfill the Objective?
    Respond with 'YES' or 'NO', followed by a brief explanation.
    """
    evaluation_result = get_completion(evaluation_prompt, model=evaluation_model)
    return evaluation_result
````

````python
def check_clarity_and_usefulness(original_prompt, llm_response, objective, evaluation_model="llama3"):
    """
    Evaluates if an LLM's response is clear and useful according to a specified objective.

    Args:
        original_prompt (str): The prompt given to the original LLM.
        llm_response (str): The response generated by the original LLM.
        objective (str): The objective describing what constitutes a clear and useful response.
        evaluation_model (str): The model to use for the evaluation.

    Returns:
        str: A string indicating whether the response was clear/useful and why.
    """
    evaluation_prompt = f"""
    You are an expert evaluator. Your task is to determine if an AI's response is clear and useful.

    Original Prompt: """{original_prompt}"""
    AI Response: """{llm_response}"""
    Objective: """{objective}"""

    Based on the above, is the AI Response clear, easy to understand, and useful in achieving the objective?
    Respond with 'YES' or 'NO', followed by a brief explanation.
    """
    evaluation_result = get_completion(evaluation_prompt, model=evaluation_model)
    return evaluation_result
````

````python
def check_iteration_quality(original_prompt, llm_response, objective, evaluation_model="llama3"):
    """
    Evaluates if an LLM's response demonstrates targeted and effective refinements based on an objective.

    Args:
        original_prompt (str): The initial prompt given to the LLM.
        llm_response (str): The refined response generated by the LLM.
        objective (str): The objective defining what constitutes targeted and effective refinements.
        evaluation_model (str): The model to use for the evaluation.

    Returns:
        str: A string indicating whether the iteration quality was met and why.
    """
    evaluation_prompt = f"""
    You are an expert evaluator. Your task is to determine if an AI's response shows targeted and effective refinements.

    Original Prompt: """{original_prompt}"""
    AI Response: """{llm_response}"""
    Objective: """{objective}"""

    Based on the above, does the AI Response demonstrate targeted and effective refinements in relation to the objective?
    Respond with 'YES' or 'NO', followed by a brief explanation.
    """
    evaluation_result = get_completion(evaluation_prompt, model=evaluation_model)
    return evaluation_result
````

````python
def check_constraint_adherence(original_prompt, llm_response, objective, evaluation_model="llama3"):
    """
    Evaluates if an LLM's response adheres to specified constraints (e.g., length, keys, tone).

    Args:
        original_prompt (str): The prompt given to the original LLM.
        llm_response (str): The response generated by the original LLM.
        objective (str): The objective defining the constraints to be adhered to.
        evaluation_model (str): The model to use for the evaluation.

    Returns:
        str: A string indicating whether the constraints were met and why.
    """
    evaluation_prompt = f"""
    You are an expert evaluator. Your task is to determine if an AI's response adheres to all specified constraints.

    Original Prompt: """{original_prompt}"""
    AI Response: """{llm_response}"""
    Objective: """{objective}"""

    Based on the above, does the AI Response completely and accurately adhere to all the constraints mentioned in the objective?
    Respond with 'YES' or 'NO', followed by a brief explanation.
    """
    evaluation_result = get_completion(evaluation_prompt, model=evaluation_model)
    return evaluation_result
````
